### Week 2

### Step1:Import Libraries for modelling 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture

from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
```

These are the libraries needed for preprocessing, clustering, dimensionality reduction, and evaluating cluster quality using multiple metrics.

### Step 2 — Create a copy of your EDA dataset

```python
model_data = data.copy()
model_data.shape
```

A separate copy of the dataset is created so that preprocessing and model-related changes do not affect the original dataset used during EDA.

### Step 3 — Handle Missing Values

```python
model_data['Income'] = model_data['Income'].fillna(
    model_data['Income'].median()
)
model_data.isnull().sum().sum()
```

Missing Income values are replaced with the median Income.
Median is used because it is less affected by extreme values than the mean.

### Step 4 — Feature Engineering

```python
# Total spending by customer
model_data['Total_Spend'] = (
    model_data['MntWines'] +
    model_data['MntFruits'] +
    model_data['MntMeatProducts'] +
    model_data['MntFishProducts'] +
    model_data['MntSweetProducts'] +
    model_data['MntGoldProds']
)

# Total purchases through different channels
model_data['Total_Purchases'] = (
    model_data['NumWebPurchases'] +
    model_data['NumCatalogPurchases'] +
    model_data['NumStorePurchases']
)

# Total children and teenagers
model_data['Total_Children'] = (
    model_data['Kidhome'] +
    model_data['Teenhome']
)

# Total accepted campaigns
model_data['Total_Campaign_Accepted'] = (
    model_data['AcceptedCmp1'] +
    model_data['AcceptedCmp2'] +
    model_data['AcceptedCmp3'] +
    model_data['AcceptedCmp4'] +
    model_data['AcceptedCmp5']
)
```


These features combine related variables into simple measures of customer behaviour.
They help the clustering models identify customers with similar spending, purchasing and promotional behaviour.
New features are created to summarize important aspects of customer behaviour such as total spending, total purchases, family size and campaign acceptance.





### Calculate customer tenure in days

```python
model_data['Dt_Customer'] = pd.to_datetime(
    model_data['Dt_Customer'],
    errors='coerce'
)

reference_date = model_data['Dt_Customer'].max()

model_data['Customer_Tenure_Days'] = (
    reference_date - model_data['Dt_Customer']
).dt.days

print("Customer tenure created successfully.")

print(
    model_data['Customer_Tenure_Days']
    .describe()
)
```

This measures how long each customer had been enrolled relative to the latest enrollment date in the dataset. It is a relative tenure measure, not necessarily the customer's true tenure at the original data snapshot date.

### Step5 — Select features

```python
features = [
    'Income',
    'Recency',
    'Total_Spend',
    'Total_Purchases',
    'NumDealsPurchases',
    'NumWebVisitsMonth',
    'Total_Children',
    'Total_Campaign_Accepted',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases',
    'Customer_Tenure_Days'
]

X = model_data[features]

X.head()
```

Relevant numerical variables are selected for clustering.
Identifier and non-behavioural variables are excluded because they do not provide useful information for forming customer segments.The selected features represent customer income, recency, spending, purchasing behaviour, family information and campaign participation.
These variables will be used as inputs to the clustering models.

### Step 6 — Check the modelling data

```python
print("Shape of modelling data:", X.shape)
print("\nMissing values:")
print(X.isnull().sum())
```

The selected dataset is checked once again to make sure that the variables used for modelling are numerical and contain no missing values.

### Step 7 — Standardize the features

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)
X_scaled.head()
```

StandardScaler transforms the features so that they have a similar scale.
This prevents variables with larger numerical values from dominating the clustering process.

### Step 8 — K-MEANS: Find suitable K

```python
from sklearn.cluster import KMeans
wcss = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    wcss,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

The Elbow Method shows how within-cluster variation changes as the number of clusters increases.
The point where the improvement starts becoming smaller can be considered as a possible number of clusters.

### Step 9 — Silhouette Score for K

```python
from sklearn.metrics import silhouette_score
silhouette_scores = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(
        X_scaled,
        labels
    )
    silhouette_scores.append(score)
for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k} : Silhouette Score = {score:.4f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.show()
```

The Silhouette Score measures how well each customer fits within its assigned cluster compared with other clusters.
A higher score generally indicates better separation between clusters.

### Step 10 — Build K-Means

```python
k = 5

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)

model_data['KMeans_Cluster'] = kmeans_labels

model_data['KMeans_Cluster'].value_counts().sort_index()
```

K-Means assigns each customer to one of five clusters based on similarity in the selected features.
Each cluster represents a potential customer segment.

### Step 11 — Evaluate K-Means

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_labels
)

kmeans_db = davies_bouldin_score(
    X_scaled,
    kmeans_labels
)

kmeans_ch = calinski_harabasz_score(
    X_scaled,
    kmeans_labels
)

print("K-Means Evaluation")
print("------------------")
print("Silhouette Score:", round(kmeans_silhouette, 4))
print("Davies-Bouldin Index:", round(kmeans_db, 4))
print("Calinski-Harabasz Score:", round(kmeans_ch, 4))
```

The K-Means model is evaluated using three internal clustering metrics.
Higher Silhouette and Calinski-Harabasz scores are preferred, while a lower Davies-Bouldin score is preferred.

### Step  12 — MODEL 2: HIERARCHICAL CLUSTERING

```python
from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=5,
    linkage='ward'
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

model_data['Hierarchical_Cluster'] = hierarchical_labels

model_data['Hierarchical_Cluster'].value_counts().sort_index()
```

The hierarchical model groups customers based on their similarity and creates five customer segments.
The resulting cluster assignments will be evaluated using the same metrics as K-Means.

### Step 12 — Evaluate Hierarchical

```python
hier_silhouette = silhouette_score(
    X_scaled,
    hierarchical_labels
)

hier_db = davies_bouldin_score(
    X_scaled,
    hierarchical_labels
)

hier_ch = calinski_harabasz_score(
    X_scaled,
    hierarchical_labels
)

print("Hierarchical Clustering Evaluation")
print("----------------------------------")
print("Silhouette Score:", round(hier_silhouette, 4))
print("Davies-Bouldin Index:", round(hier_db, 4))
print("Calinski-Harabasz Score:", round(hier_ch, 4))
```

The hierarchical clustering results are evaluated using Silhouette Score, Davies-Bouldin Index and Calinski-Harabasz Score.
These results will later be compared with the other clustering models.

### Step 13 — Dendrogram

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linkage_matrix = linkage(
    X_scaled,
    method='ward'
)

plt.figure(figsize=(12, 6))

dendrogram(
    linkage_matrix,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers / Groups")
plt.ylabel("Distance")

plt.show()
```

The dendrogram shows how observations are progressively combined into groups.
It provides a visual representation of the hierarchical structure of the customer data.

### Step 14 — MODEL 3: DBSCAN

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(X_scaled)

model_data['DBSCAN_Cluster'] = dbscan_labels

model_data['DBSCAN_Cluster'].value_counts().sort_index()
```

DBSCAN groups customers based on the density of nearby observations.
Customers that do not belong to a sufficiently dense region are identified as noise.

### Step 15 — Evaluate DBSCAN

```python
mask = dbscan_labels != -1

n_clusters_dbscan = len(
    set(dbscan_labels)
) - (1 if -1 in dbscan_labels else 0)

print("Number of clusters:", n_clusters_dbscan)
print("Number of noise points:", (~mask).sum())
```

```python
if n_clusters_dbscan >= 2:
    
    dbscan_silhouette = silhouette_score(
        X_scaled[mask],
        dbscan_labels[mask]
    )
    
    dbscan_db = davies_bouldin_score(
        X_scaled[mask],
        dbscan_labels[mask]
    )
    
    dbscan_ch = calinski_harabasz_score(
        X_scaled[mask],
        dbscan_labels[mask]
    )
    
    print("Silhouette Score:", round(dbscan_silhouette, 4))
    print("Davies-Bouldin Index:", round(dbscan_db, 4))
    print("Calinski-Harabasz Score:", round(dbscan_ch, 4))
    
else:
    print("DBSCAN did not produce enough clusters for evaluation.")
```

The GMM model is evaluated using the same clustering metrics used for the other models.
Using common metrics allows the four models to be compared fairly.

```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=5,
    random_state=42
)

gmm_labels = gmm.fit_predict(X_scaled)

model_data['GMM_Cluster'] = gmm_labels

model_data['GMM_Cluster'].value_counts().sort_index()
```

```python
gmm_silhouette = silhouette_score(
    X_scaled,
    gmm_labels
)

gmm_db = davies_bouldin_score(
    X_scaled,
    gmm_labels
)

gmm_ch = calinski_harabasz_score(
    X_scaled,
    gmm_labels
)

print("GMM Evaluation")
print("----------------")
print("Silhouette Score:", round(gmm_silhouette, 4))
print("Davies-Bouldin Index:", round(gmm_db, 4))
print("Calinski-Harabasz Score:", round(gmm_ch, 4))
```

### Step 19 - Compare ALL FOUR MODELS

```python
results = pd.DataFrame({
    'Model': [
        'K-Means',
        'Hierarchical',
        'DBSCAN',
        'GMM'
    ],
    'Silhouette Score': [
        kmeans_silhouette,
        hier_silhouette,
        dbscan_silhouette,
        gmm_silhouette
    ],
    'Davies-Bouldin Index': [
        kmeans_db,
        hier_db,
        dbscan_db,
        gmm_db
    ],
    'Calinski-Harabasz Score': [
        kmeans_ch,
        hier_ch,
        dbscan_ch,
        gmm_ch
    ]
})

results = results.round(4)

results
```

The four clustering algorithms are compared using common evaluation metrics.
This comparison helps identify the model that provides a suitable balance between cluster quality and practical customer segmentation.

### Step 20 — Sort the models

```python
ranked_results = results.sort_values(
    by='Silhouette Score',
    ascending=False,
    na_position='last'
)

ranked_results
```

The models are ranked according to their Silhouette Score.
However, the final model is not selected using one metric alone; Davies-Bouldin Index, Calinski-Harabasz Score and cluster distribution are also considered.
